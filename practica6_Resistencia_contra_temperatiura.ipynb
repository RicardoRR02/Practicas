```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import linregress
```

```python
data=pd.read_csv('remten2.csv')
V=data['V'][0]
T=data['T1']
I=data['I1']
T=np.array(T)
T=T+273.15
I=np.array(I)*10**(-3)
R=[V/i for i in I]
print(V)
```

```python

plt.plot(T,I)
plt.title("Mediciones de la corriente",fontsize=16,fontweight='bold')
plt.ylabel("I (A)", fontsize=14)
plt.xlabel(r"T $(K)$", fontsize=14)
plt.grid()
plt.show()
plt.plot(T,R)
plt.title("Resistencia en función de la temperatura",fontsize=16,fontweight='bold')
plt.ylabel(r"R ($\Omega$)", fontsize=14)
plt.xlabel(r"T $(K)$", fontsize=14)
plt.grid()
plt.show()
```

```python
lnr=np.log(R)
tinv=[1/i for i in T]
plt.plot(tinv,lnr)
plt.title("Gráfica del modelo esperado",fontsize=16,fontweight='bold')
plt.xlabel('1/T (1/K)')
plt.ylabel(r'log(R) (log($\Omega$))')
plt.grid()
plt.show()
```

```python
re=linregress(tinv,lnr)
t=np.linspace(min(tinv),max(tinv))
plt.plot(tinv,lnr,'o')
plt.plot(t,re.intercept+re.slope*t,'r')
plt.title("Ajuste del modelo esperado.",fontsize=16,fontweight='bold')
plt.xlabel('1/T (1/K)')
plt.ylabel(r'log(R) (log($\Omega$))')
plt.grid()
plt.show()
print(re)
```

```python
N=2
for j in range(0,N):
  if j==0:
    temporal1=[]
    temporal2=[]
    for i in range(0,len(T)-1):
      if lnr[i+1]>=lnr[i] and abs(lnr[i+1]-lnr[i])<10**(-3):
        temporal1.append(lnr[i])
        temporal2.append(tinv[i])
    temporal1.append(lnr[len(T)-1])
    temporal2.append(tinv[len(T)-1])
  else:
    temporal3=[]
    temporal4=[]
    for i in range(0,len(temporal1)-1):
      if temporal1[i+1]>=temporal1[i] and abs(temporal1[i+1]-temporal1[i])<10**(-3):
        temporal3.append(temporal1[i])
        temporal4.append(temporal2[i])
    temporal3.append(temporal1[len(temporal1)-1])
    temporal4.append(temporal2[len(temporal2)-1])
    temporal1=temporal3
    temporal2=temporal4
tinfiltrado=temporal2
lnfiltrado=temporal1
redos=linregress(tinfiltrado,lnfiltrado)
t=np.linspace(min(tinfiltrado),max(tinfiltrado))
plt.plot(t,redos.intercept+redos.slope*t,'r')
plt.scatter(tinfiltrado,lnfiltrado)
plt.title("Ajuste y filtrado del modelo esperado.",fontsize=16,fontweight='bold')
plt.xlabel('1/T (1/K)')
plt.ylabel(r'log(R) (log($\Omega$))')
plt.grid()
plt.show()
print(redos)
```
