```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import math as ma
data=pd.read_csv('FH.csv')
#print(data.head)
run1=data["Run 1"]
run2=data["Run 2"]
run3=data["Run 3"]
run4=data["Run 4 "]

runprom=[]
var=[]
for i in range(0,60):
  valor=(run1[i]+run2[i]+run3[i]+run4[i])/4
  runprom.append(valor)
  vari=ma.sqrt((run1[i]-valor)**2+(run2[i]-valor)**2+(run3[i]-valor)**2+(run4[i]-valor)**2)/ma.sqrt(3)
  var.append(vari)
```

```python
plt.figure(figsize=(10,10))
plt.plot(range(1,61),runprom,linewidth=3)
plt.errorbar(range(1,61),runprom,yerr=var,fmt='o',color='red',capsize=5,ecolor='black')
plt.title("Valores promedio",fontsize=16,fontweight='bold')
plt.ylabel("Corriente (nA)", fontsize=14)
plt.xlabel("Voltaje (V)", fontsize=14)
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.plot(range(1,61),run1,color='red',linewidth=3)
plt.plot(range(1,61),run2,linestyle='-',color='blue',linewidth=3)
plt.plot(range(1,61),run3,linestyle=':',color='purple',linewidth=3)
plt.plot(range(1,61),run4,linestyle='--',color='green',linewidth=3)
plt.title("Valores medidos",fontsize=16,fontweight='bold')
plt.ylabel("Corriente (nA)", fontsize=14)
plt.xlabel("Voltaje (V)", fontsize=14)
plt.legend(['Medición 1','Medición 2','Medición 3','Medición 4'])
plt.grid()
plt.show()
```
