```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import linregress
```

```python
data=pd.read_csv('beer.csv')
d=data['d']*10
n=list(range(0,11))
x=[]
dx=[]
for i in range(1,11):
    x.append(np.mean(data.iloc[:,i]))
    dx.append(np.std(data.iloc[:,i]))
print(len(x),len(dx),len(d),len(data))
x=np.array(x)
dx=np.array(dx)
```

```python
plt.errorbar(d,x,yerr=dx,fmt='o')
plt.title("Diferencial de potencial de las muestras",fontsize=16,fontweight='bold')
plt.ylabel(r"V (mV)", fontsize=14)
plt.xlabel(r"d (mm)", fontsize=14)
plt.grid()
```

```python
x=x*10**(-3)
dx=dx*10**(-3)
lnx=np.log(x)
#lnx=lnx-lnx[0]
plt.plot(d,lnx,'o')
plt.title("Aplicación del modelo",fontsize=16,fontweight='bold')
plt.ylabel(r"ln(V) (mV)", fontsize=14)
plt.xlabel(r"d (mm)", fontsize=14)
plt.grid()
```

```python
d=d*10**(-3)
lnv=lnx+np.log(10**(-3))
res=linregress(d,lnv)
plt.plot(d,lnv,'o')
plt.plot(d,res.intercept+res.slope*d,'-',color='red')
plt.title("Regresión lineal y aplicaión del modelo",fontsize=16,fontweight='bold')
plt.legend(['Datos','Regresión lineal'])
plt.ylabel(r"ln(V) (V)", fontsize=14)
plt.xlabel(r"d (m)", fontsize=14)
plt.grid()
print(res)
```
